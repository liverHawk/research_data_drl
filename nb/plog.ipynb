```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../result.csv")
```

```python
df
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# 混淆矩阵
cm = confusion_matrix(df['actual'], df['action'])
print(type(cm))

same_shape_matrix = np.zeros(cm.shape)

# average for column
for i in range(cm.shape[1]):
    print(cm[:, i].sum())
    if cm[:, i].sum() != 0:
        print(cm[:, i])
        same_shape_matrix[:, i] = cm[:, i] / float(cm[:, i].sum())
        print(same_shape_matrix[:, i])
print(same_shape_matrix)
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(same_shape_matrix, annot=True, fmt='.3f', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
