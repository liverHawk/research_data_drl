```python
import numpy as np
import pandas as pd
from glob import glob
```

```python
files = glob("../data/train/raw/*.csv.gz")
dfs = [pd.read_csv(file) for file in files]
df = pd.concat(dfs)
```

```python
value_counts = df["Label"].value_counts()

except_label = value_counts.index[:2]
except_df = df[df["Label"].isin(except_label)]
sampling_df = df[~df["Label"].isin(except_label)]
```

```python
import over_sampling
sampling_props = over_sampling.SamplingProps(sampling_df, "Label")
resample_df, props = over_sampling.random(sampling_props)
```
